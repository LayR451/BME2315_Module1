* ### Societal determinants
    Certain lifestyle and background factors can influence an individual's likelihood to develop Alzhimer's disease later in life. Many of these factors include education level, access to healthcare, and phyiscal environment in one's community, as well as someone's demographics. Some studies have shown an association between lower levels of education and development of Alzheimer's and dementia, with cognitive decline among adults age 45 and older being twice as high for people without a high school diploma. Some scientists believe that education and knowledge levels can be stored or "banked" in early years of life, a phenomenon called "cognitive reserve". Additionally, many people across the United States do not have access to adequate healthcare, reducing chances of early diagnosis and ease of treatment. Social isolation or loneliness has also been found to increase risk of developing Alzheimer's disease and dementia. These factors ultimately demonstrate the significant roles that lifestyle, background, and geography can play when it comes to affecting someone's risk of developing Alzheimer's.

* ### Symptoms

* ### Diagnosis
    Early symptoms of Alzheimer's disease include memory loss, trouble concentrating, poor judgement and decision making, social withdrawl, changes in mood or personality, trouble with language skills, and trouble with routine daily tasks. To diagnose Alzheimer's, healthcare providers perform a series of tests including mental status testing, neurophychological tests, labrotory tests for certain proteins (specifically p-tau217 and amyloid proteins), and brain imaging tests such as an MRI, CT, or PET scan. Neurophysiologists and neurologists are typically responsible for testing and evaluating patients for Alzheimer's diagnosis. Early diagnosis can help patients begin therapies, medications, and clinical trials much earlier, reducing or slowing the impact of the disease. 

    Source:

Some insurances cover the cost of long term stays, and others do not. As for medications, a variety exist to ease some Alzheimer's symptoms, and are largely covered by most major insurance providers, including Medicare and Medicaid.

* ### Disease progression & prognosis
    Progression of Alzheimer’s disease happens in several stages. In the early stage, patients may function independently but may display some symptoms, including having issues with memory cognition. Middle stage is often the longest stage for most patients, and lasts several years. During this stage, symptoms of dementia are much more present, and patients require a higher level of care. Damage to nerve cells can make it difficult for patients to express thoughts and perform daily tasks independently. Symptoms of middle stage Alzheimer’s include being unable to recall important information such as their address or phone number, having trouble controlling their bladder and bowels, experiencing changes in sleep patterns, and demonstrating personality or behavioral changes. Many patients in this stage are still able to participate in daily activities, but require assistance. The final stage of Alzheimer's is the late stage, where patients have worsened symptoms and require around the clock care. Patients may have difficulty doing basic tasks, such as eating, walking, sitting, and swallowing. They may also become more venerable to infections.

* ### Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
    The main characteristic of Alzheimer’s disease includes the extracellular buildup of amyloid-beta plaques and tau protein tangles in the intracellular environment. The accumulation of these compounds in the brain can result in neuron death and loss of synapses, which can contribute to progressive cognitive decline. Additionally, newer research has demonstrated that Alzheimer’s disease has been shown to alter glutamate signaling, the primary neurotransmitter in the Central Nervous System. This results in chronic activation of NMDA receptors due to potential glutamate toxicity, which has been shown to lead to neurodegeneration. Many different parts of the brain are susceptible to this damage, including the brain stem, amygdala, hypothalamus, and regions of the cerebral cortex, and others. This directly results in a decline in the ability of the brain to function and causes common symptoms like memory loss and having trouble thinking, speaking, and completing actions.

*Step 2: Outlier Analysis*

After our initial assessing of the data, we determined that an outlier analysis was necessary. We did a traditional Inter Quartile range check for values more than 1.5 times away from the range, and found 5 values here. Then, to further consider them, we ran a Z test to see how many standard deviations those data points were away from the mean. Z values of greater than 3 were eliminated. This meant that we removed 1 outlier. 

```python
# First check of outlier analysis! 
df = pd.read_csv("Metadata and Protein Data for Module 1.csv")
col = "ABeta42 pg/ug"       # protein concentration column to check
id_col = "Donor ID"        # column that identifies each patient
df = df.dropna(subset=[col])

# IQR: flag anything beyond 1.5 * IQR from the quartiles
q1, q3 = df[col].quantile([0.25, 0.75])
iqr = q3 - q1
low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
df["outlier"] = (df[col] < low) | (df[col] > high)

# z-score 
df["z"] = (df[col] - df[col].mean()) / df[col].std()

print(f"Normal range: {low:.2f} to {high:.2f}")
print(df.loc[df["outlier"], [id_col, col, "z"]])

df.boxplot(column=col)
plt.show()
```

*Step 3: The data was compiled into graphs with the desired traits*

And 

*Step 4: The data was analysed for the possibility of significance*

Python is then used to calculate the duration of the disease

```python
# Calculate disease duration
# Disease duration = age at death - age of symptom onset

disease_duration = []

for patient in Patient.all_patients:
    if patient.age_onset is not None:
        duration = patient.age_at_death - patient.age_onset
        disease_duration.append(duration)

# Print disease duration information
print("\nDisease duration:")
print(f"Number of patients with disease duration data = {len(disease_duration)}")
if disease_duration:
    print(f"Mean disease duration = {statistics.mean(disease_duration):.2f} years")
    print(f"Standard deviation of disease duration = {statistics.stdev(disease_duration):.2f} years")
else:
    print("Mean disease duration = N/A")
    print("Standard deviation of disease duration = N/A")
```

```python
# Now, we will create graphs. 
# First up, is the bar graph: 


# Bar graph
# Mean disease duration +/- standard deviation for each education level
high_school_duration = []
trade_school_duration = []
bachelors_duration = []
graduate_duration = []
professional_duration = []

for patient in Patient.all_patients:
    if patient.age_dementia_diagnosis is not None:
        duration = patient.age_at_death - patient.age_dementia_diagnosis

        if patient.highest_education == "High School":
            high_school_duration.append(duration)
        elif patient.highest_education == "Trade School/ Tech School":
            trade_school_duration.append(duration)
        elif patient.highest_education == "Bachelors":
            bachelors_duration.append(duration)
        elif patient.highest_education == "Graduate (PhD/Masters)":
            graduate_duration.append(duration)
        elif patient.highest_education == "Professional":
            professional_duration.append(duration)


def safe_mean(values):
    return statistics.mean(values) if values else 0.0


def safe_stdev(values):
    return statistics.stdev(values) if len(values) > 1 else 0.0

# Calculate means
mean_high_school = safe_mean(high_school_duration)
mean_trade_school = safe_mean(trade_school_duration)
mean_bachelors = safe_mean(bachelors_duration)
mean_graduate = safe_mean(graduate_duration)
mean_professional = safe_mean(professional_duration)

# Calculate standard deviations
stdev_high_school = safe_stdev(high_school_duration)
stdev_trade_school = safe_stdev(trade_school_duration)
stdev_bachelors = safe_stdev(bachelors_duration)
stdev_graduate = safe_stdev(graduate_duration)
stdev_professional = safe_stdev(professional_duration)

# Print statistics
print("\nDisease duration by education level:")
print(f"High School: mean = {mean_high_school:.2f}, SD = {stdev_high_school:.2f}")
print(f"Trade School: mean = {mean_trade_school:.2f}, SD = {stdev_trade_school:.2f}")
print(f"Bachelors: mean = {mean_bachelors:.2f}, SD = {stdev_bachelors:.2f}")
print(f"Graduate: mean = {mean_graduate:.2f}, SD = {stdev_graduate:.2f}")
print(f"Professional: mean = {mean_professional:.2f}, SD = {stdev_professional:.2f}")

# Make bar graph

education_labels = [
    "High School",
    "Trade School",
    "Bachelors",
    "Graduate",
    "Professional",
]

education_means = [
    mean_high_school,
    mean_trade_school,
    mean_bachelors,
    mean_graduate,
    mean_professional,
]

education_stdev = [
    stdev_high_school,
    stdev_trade_school,
    stdev_bachelors,
    stdev_graduate,
    stdev_professional,
]

# Labels to make the bar graph more readable


plt.bar(
    education_labels,
    education_means,
    yerr=education_stdev,
    capsize=5
)

plt.title(
    "Mean Disease Duration by Highest Education Level"
)

plt.xlabel(
    "Highest Education Level"
)

plt.ylabel(
    "Mean Disease Duration (years)"
)

plt.xticks(rotation=20)

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.5
)

plt.tight_layout()
plt.show()


```

```python
# Scatter plot
# Years of education vs. age of onset of cognitive symptoms

years_education = []
age_onset = []

for patient in Patient.all_patients:
    if patient.age_onset is not None and patient.years_education is not None:
        years_education.append(patient.years_education)
        age_onset.append(patient.age_onset)


model = LinearRegression()
model.fit(np.array(years_education).reshape(-1, 1), np.array(age_onset).reshape(-1, 1))

X = np.linspace(
    min(years_education),
    max(years_education),
    100
).reshape(-1, 1)

# Make scatter plot
plt.scatter(
    years_education,
    age_onset
)

plt.title(
    "Years of Education vs. Age of Onset of Cognitive Symptoms"
)

plt.xlabel(
    "Years of Education (years)"
)

plt.ylabel(
    "Age of Onset of Cognitive Symptoms (years)"
)

plt.grid(
    linestyle="--",
    alpha=0.5
)
# Add regression line
plt.plot(X, model.predict(X).ravel(), color="red", linewidth=2)

plt.tight_layout()
plt.show()

# make r squared value
r_squared = model.score(np.array(years_education).reshape(-1, 1), np.array(age_onset).reshape(-1, 1))
print(f"R-squared value: {r_squared:.2f}")

# Calculate p-value
slope, intercept, r_value, p_value, standard_error = stats.linregress(
    years_education,
    age_onset
)

print(f"P-value = {p_value:.3f}")
```

```python

# This scatterplot relates length of time from disease onset to death to the amount of amyloid plaque

# These are empty lists to store the data
disease_length = []
amyloid_plaque_levels = []

for p in Patient.all_patients:
    if p.age_onset is not None and p.age_at_death is not None:
        disease_length.append(p.age_at_death - p.age_onset)
        amyloid_plaque_levels.append(p.abeta42)

x_values = np.array(disease_length, dtype=float)
y_values = np.array(amyloid_plaque_levels, dtype=float)

# Uses patients with valid values for both variables 
valid_data = np.isfinite(x_values) & np.isfinite(y_values)
x = x_values[valid_data].reshape(-1, 1)
y = y_values[valid_data].reshape(-1, 1)

# Make a linear regression
model = LinearRegression()
model.fit(x, y)
r2 = model.score(x, y)

# Plot and show
plt.scatter(x, y, color='blue')
plt.plot(x, model.predict(x), color='red')
plt.xlabel('Length of Disease (years)')
plt.ylabel('Amyloid Plaque (pg/µg)')
plt.title('Scatter Plot of Disease Length vs Amyloid Plaque')
plt.show()

# R and p values
print("The R squared value is:", r2)
r, p_val = stats.pearsonr(x, y)
print('The p value is:', p_val)
```
