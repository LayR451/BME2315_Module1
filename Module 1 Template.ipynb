The economic burden of Alzheimer's disease is very significant, as diagnosed patients often require substantial accommodations, treatments, and often must live in assisted living and/or memory care facilities. According to Nature, the total cost of care for someone with Alzheimer's disease and related dementia (ADRD) was upwards of $28,000 per year in 2016, with total costs for ADRD patients in the United States being in the hundreds of billions (about $355 billion) when additional factors like forgone wages are accounted for. Majority of these costs are paid for by Medicate and Medicaid Services, with up to 67% of total costs (or 1 out of every 5 dollars of Medicare funding) is spent on the treatment and care of people with Alzheimer's. This makes Alzheimer’s one of the most expensive diseases in America, totaling more than treatments for cancer or heart disease in the last 5 years of life.
