* ### Economic burden
    The economic burden of Alzheimer's disease is very significant, as diagnosed patients often require substantial accomidations, treatments, and often must live in assisted living and/or memory care facillities. According to Nature, the total cost of care for someone with Alzheimer's disease and related dimentia (ADRD) was upwards of $28,000 per year in 2016, with total costs for ADRD patients in teh United States being in the hundreds of billions (about $355 billion) when additional factors like forgone wages are accounted for. Majority of these costs are paid for by Medicare and Medicaid Services, with up to 67% of total costs (or 1 out of every 5 dollars of medicare funding) is spent on the treatment and care of people with Alzheimer's. This makes Alzheimer's care and treatment some of the most expensive in America, totaling more than treatments for cancer or heart disease in the last 5 years of life.

* #### Risk factors (genetic, lifestyle) 
    Some risk factors that have been identified is cognitive slowing or lack of mental stimulation in later life, as well as poor diet and exercise. There are also links to the ApoE gene

* ### Societal determinants
    Certain lifestyle and background factors can influence an individual's likelihood to develop Alzhimer's disease later in life. Many of these factors include education level, access to healthcare, and phyiscal environment in one's community, as well as someone's demographics. Some studies have shown an association between lower levels of education and development of Alzheimer's and dementia, with cognitive decline among adults age 45 and older being twice as high for people without a high school diploma. Some scientists believe that education and knowledge levels can be stored or "banked" in early years of life, a phenomenon called "cognitive reserve". Additionally, many people across the United States do not have access to adequate healthcare, reducing chances of early diagnosis and ease of treatment. Social isolation or loneliness has also been found to increase risk of developing Alzheimer's disease and dementia. Finally, specific demographics can increase someone's risk, as Women have an approximately 17% higher risk for developing Alzheimer's than men. Additionally, African Americans and Native Hawaiians are at highest risk for developing Alzheimer's, while Japanese Americans and Fillipinos were at a lower risk. These factors ultimately demonstrate the significant roles that lifestyle, background, and geography can play when it comes to affecting someone's risk of developing Alzheimer's.

* ### Symptoms
    Alzheimer's symptoms include a range of things from memory loss, challenges with daily taks, lack of spatial processing, issues with time-keeping, mood changes and personaility changes. These symptoms can vary day to day, often creating a sense of "good and bad days". Alzheimer's tends to develop into motor issues, including gait issues, poor balance, and eventually inability to swallow.

* ### Diagnosis
    Early symptoms of Alzheimer's disease include memory loss, trouble concentrating, poor judgement and decision making, social withdrawl, changes in mood or personality, trouble with language skills, and trouble with routine daily tasks. To diagnose Alzheimer's, healthcare providers perform a series of tests including mental status testing, neurophychological tests, labrotory tests for certain proteins (specifically p-tau217 and amyloid proteins), and brain imaging tests such as an MRI, CT, or PET scan. Neurophysiologists and neurologists are typically responsible for testing and evaluating patients for Alzheimer's diagnosis. Early diagnosis can help patients begin therapies, medications, and clinical trials much earlier, reducing or slowing the impact of the disease. 

    Source:
    https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/in-depth/alzheimers/art-20048075 

* ### Standard of care treatments (& reimbursement) 
    Alzheimer's unfornuately has no cure as of right now. Instead, most individuals are cared for by a family member, child, spouse or hired caregiver. Family member care may not invoke monetary expense, but is linked with increased anxiety and decreased life quality for the caregiver. Because Alzheimer's devolves over time, they often have an increase in care needs as the disease progresses. Many individuals will end up in specialized long term care programs. For patients with Medicare, this will be a very expensive out of pocket cost for their families, while Medicaid does cover long term care stays. Some insurances cover the cost of long term stays, and others do not. Similarly, in home care to assist with house chores, feeding, bathing and other supplimental help to the main caregiver is not covered by Medicare either, and largely variable as to if it is covered by insurances of Medicaid. As for medications, a variety exist to ease some Alzheimer's symptoms, and are largely covered by most major insurance providers, including Medicare and Medicaid. 

* ### Disease progression & prognosis
    Progression of Alzheimer’s disease happens in several stages. In the early stage, patients may function independently but may display some symptoms, including having issues with memory cognition. Middle stage is often the longest stage for most patients, and lasts several years. During this stage, symptoms of dementia are much more present, and patients require a higher level of care. Damage to nerve cells can make it difficult for patients to express thoughts and perform daily tasks independently. Symptoms of middle stage Alzheimer’s include being unable to recall important information such as their address or phone number, having trouble controlling their bladder and bowels, experiencing changes in sleep patterns, and demonstrating personality or behavioral changes. Many patients in this stage are still able to participate in daily activities, but require assistance. The final stage of Alzheimer's is the late stage, where patients have worsened symptoms and require around the clock care. Patients may have difficulty doing basic tasks, such as eating, walking, sitting, and swallowing. They may also become more venerable to infections. 

    Source:
    https://www.alz.org/alzheimers-dementia/stages 

* ### Continuum of care providers
    Care for patients with Alzheimer's tends to involve a general care provider, then often a geriatric specialist or maybe straigyht to the nuerologist. From diagnosis onwards, routine appointments may occur with a nuerologists, and the pateint will reciee most of their care from a home care provider, a pharmacists, or a care facility. 

* ### Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)
    The main characteristic of Alzheimer’s disease includes the extracellular buildup of amyloid-beta plaques and tau protein tangles in the intracellular environment. The accumulation of these compounds in the brain can result in neuron death and loss of synapses, which can contribute to progressive cognitive decline. Additionally, newer research has demonstrated that Alzheimer’s disease has been shown to alter glutamate signaling, the primary neurotransmitter in the Central Nervous System. This results in chronic activation of NMDA receptors due to potential glutamate toxicity, which has been shown to lead to neurodegeneration. Many different parts of the brain are susceptible to this damage, including the brain stem, amygdala, hypothalamus, and regions of the cerebral cortex, and others. This directly results in a decline in the ability of the brain to function and causes common symptoms like memory loss and having trouble thinking, speaking, and completing actions.  

    Sources:
    https://pmc.ncbi.nlm.nih.gov/articles/PMC5290713/ 
    
* ### Clinical Trials/next-gen therapies
    Alzheimer's research is at the forefront of many researcher's minds, and as such, many new avenues of treatment have arisen in recent years. To start, blood screening options are now being more widely avabile to check for Alzheimer's years or even decades before symptoms begin. Another possible therapy being considered is lithium remediation via dietary suppliments, as a decerease in lithium is shown to accelerate the disease progession, and is common in patients as amyloid plaque uses lithium stores. Another example is GRK2 proteins and Compound 10. GRK2 become inactive and harmful in Alzheimer's, and so some current experimenetal models are showing that by making Compound 10 bind to GRK2, it slows the nerve cell death and therefore increased life span. However, treatments like these require significant long term trials and FDA testing, and so the timeline to be avaible on the market is likely long.

Python is then used to calculate the duraiton of the disease

```python
# Calculate disease duration
# Disease duration = age at death - age at dementia diagnosis

disease_duration = []

for patient in Patient.all_patients:
    if patient.age_dementia_diagnosis is not None:
        duration = patient.age_at_death - patient.age_dementia_diagnosis
        disease_duration.append(duration)

# Print disease duration information
print("\nDisease duration:")
print(f"Number of patients with disease duration data = {len(disease_duration)}")
if disease_duration:
    print(f"Mean disease duration = {statistics.mean(disease_duration):.2f} years")
    print(f"Standard deviation of disease duration = {statistics.stdev(disease_duration):.2f} years")
else:
    print("Mean disease duration = N/A")
    print("Standard deviation of disease duration = N/A")
```

Next, we find Amyloid Beta - 40 and Amyloid Beta - 42 levels from the dataset

```python
# Extract Amyloid-Beta 40 and Amyloid-Beta 42 levels

abeta40_levels = []
abeta42_levels = []

for patient in Patient.all_patients:
    if patient.abeta40 is not None:
        abeta40_levels.append(patient.abeta40)

    if patient.abeta42 is not None:
        abeta42_levels.append(patient.abeta42)

# Print Amyloid-Beta information
print("\nAmyloid-Beta 40:")
print(f"Number of patients with Amyloid-Beta 40 data = {len(abeta40_levels)}")

if abeta40_levels:
    print(f"Mean Amyloid-Beta 40 = {statistics.mean(abeta40_levels):.2f} pg/ug")
    print(f"Standard deviation = {statistics.stdev(abeta40_levels):.2f} pg/ug")

print("\nAmyloid-Beta 42:")
print(f"Number of patients with Amyloid-Beta 42 data = {len(abeta42_levels)}")

if abeta42_levels:
    print(f"Mean Amyloid-Beta 42 = {statistics.mean(abeta42_levels):.2f} pg/ug")
    print(f"Standard deviation = {statistics.stdev(abeta42_levels):.2f} pg/ug")
```

```python
disease_duration_abeta40 = []
abeta40_for_duration = []

disease_duration_abeta42 = []
abeta42_for_duration = []

for patient in Patient.all_patients:

    if patient.age_dementia_diagnosis is not None and patient.abeta40 is not None:
        duration = patient.age_at_death - patient.age_dementia_diagnosis
        disease_duration_abeta40.append(duration)
        abeta40_for_duration.append(patient.abeta40)

    if patient.age_dementia_diagnosis is not None and patient.abeta42 is not None:
        duration = patient.age_at_death - patient.age_dementia_diagnosis
        disease_duration_abeta42.append(duration)
        abeta42_for_duration.append(patient.abeta42)
```

Next, we find the lienar regression and correlation for amyloid beta 40 and amyloid beta 42 levels seperately 

```python
# Linear regression and correlation for Amyloid-Beta 40

abeta40_results = stats.linregress(
    disease_duration_abeta40,
    abeta40_for_duration
)

print("\nAmyloid-Beta 40 vs Disease Duration")
print(f"Slope = {abeta40_results.slope:.4f}")
print(f"Intercept = {abeta40_results.intercept:.4f}")
print(f"r value = {abeta40_results.rvalue:.4f}")
print(f"p value = {abeta40_results.pvalue:.4f}")


# Linear regression and correlation for Amyloid-Beta 42

abeta42_results = stats.linregress(
    disease_duration_abeta42,
    abeta42_for_duration
)

print("\nAmyloid-Beta 42 vs Disease Duration")
print(f"Slope = {abeta42_results.slope:.4f}")
print(f"Intercept = {abeta42_results.intercept:.4f}")
print(f"r value = {abeta42_results.rvalue:.4f}")
print(f"p value = {abeta42_results.pvalue:.4f}")
```
